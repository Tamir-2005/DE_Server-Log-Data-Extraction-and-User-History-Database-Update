```python
import re
from datetime import datetime

# Regex patterns
email_pattern = re.compile(r"[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.[a-z]{2,}")
date_pattern = re.compile(r"\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}")  # YYYY-MM-DD HH:MM:SS

extracted_data = []

with open("server.log", "r") as logfile:
    for line in logfile:
        email_match = email_pattern.search(line)
        date_match = date_pattern.search(line)
        if email_match and date_match:
            # Normalize date format
            raw_date = date_match.group()
            try:
                parsed_date = datetime.strptime(raw_date, "%Y-%m-%d %H:%M:%S")
                formatted_date = parsed_date.strftime("%Y-%m-%d %H:%M:%S")
            except ValueError:
                formatted_date = raw_date  # fallback if format is unknown

            # Append structured record
            extracted_data.append({
                "email": email_match.group().lower().strip(),  # clean email
                "date": formatted_date
            })

# Show transformed data
for record in extracted_data:
    print(record)
```

```python
import pymysql

# Connect to MySQL
conn = pymysql.connect(
    host="localhost",
    user="root",
    password="2103",  # replace with your MySQL password
    database="log_database",
    port=3306
)
cursor = conn.cursor()

# List all unique email addresses
print("\n--- Unique Email Addresses ---")
cursor.execute("SELECT DISTINCT email FROM user_history;")
for row in cursor.fetchall():
    print(row[0])

#  Count the number of emails received per day
print("\n--- Emails Received Per Day ---")
cursor.execute("""
SELECT DATE(log_date) AS day, COUNT(*) AS email_count
FROM user_history
GROUP BY day
ORDER BY day;
""")
for row in cursor.fetchall():
    print(f"{row[0]} → {row[1]} emails")

#  Find the first and last email date for each email address
print("\n--- First & Last Email Date Per Address ---")
cursor.execute("""
SELECT email, MIN(log_date) AS first_seen, MAX(log_date) AS last_seen
FROM user_history
GROUP BY email;
""")
for row in cursor.fetchall():
    print(f"{row[0]} → First: {row[1]}, Last: {row[2]}")

#  Count the total number of emails from each domain
print("\n--- Email Count Per Domain ---")
cursor.execute("""
SELECT SUBSTRING_INDEX(email, '@', -1) AS domain, COUNT(*) AS total_emails
FROM user_history
GROUP BY domain
ORDER BY total_emails DESC;
""")
for row in cursor.fetchall():
    print(f"{row[0]} → {row[1]} emails")

#  Close connection
cursor.close()
conn.close()
```

```python
import pymysql

# Connect to MySQL
conn = pymysql.connect(
    host="localhost",
    user="root",
    password="2103",  # replace with your MySQL password
    database="log_database",
    port=3306
)
cursor = conn.cursor()

# Helper function to run queries
def run_query(query, description):
    print(f"\n--- {description} ---")
    cursor.execute(query)
    for row in cursor.fetchall():
        print(row)

# 1. Unique email addresses
run_query("SELECT DISTINCT email FROM user_history;", "Unique Email Addresses")

# 2. Total number of emails
run_query("SELECT COUNT(*) AS total_emails FROM user_history;", "Total Emails")

# 3. Emails per day
run_query("""
SELECT DATE(log_date) AS day, COUNT(*) AS email_count
FROM user_history
GROUP BY day
ORDER BY day;
""", "Emails Per Day")

# 4. First & last email date for each email
run_query("""
SELECT email, MIN(log_date) AS first_seen, MAX(log_date) AS last_seen
FROM user_history
GROUP BY email;
""", "First & Last Email Date Per Address")

# 5. Most active email
run_query("""
SELECT email, COUNT(*) AS total_emails
FROM user_history
GROUP BY email
ORDER BY total_emails DESC
LIMIT 1;
""", "Most Active Email Address")

# 6. Emails by domain
run_query("""
SELECT SUBSTRING_INDEX(email, '@', -1) AS domain, COUNT(*) AS total_emails
FROM user_history
GROUP BY domain
ORDER BY total_emails DESC;
""", "Emails Per Domain")

# 7. Emails with more than 5 entries
run_query("""
SELECT email, COUNT(*) AS total_entries
FROM user_history
GROUP BY email
HAVING COUNT(*) > 5
ORDER BY total_entries DESC;
""", "Emails With More Than 5 Entries")

# 8. Emails in last 7 days
run_query("""
SELECT COUNT(*) AS last_7_days_emails
FROM user_history
WHERE log_date >= NOW() - INTERVAL 7 DAY;
""", "Emails in Last 7 Days")

# 9. Earliest log entry
run_query("""
SELECT email, log_date
FROM user_history
ORDER BY log_date ASC
LIMIT 1;
""", "Earliest Log Entry")

# 10. Latest log entry
run_query("""
SELECT email, log_date
FROM user_history
ORDER BY log_date DESC
LIMIT 1;
""", "Latest Log Entry")

# Close connection
cursor.close()
conn.close()
```
